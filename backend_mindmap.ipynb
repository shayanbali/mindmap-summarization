```python
from sklearn.metrics.pairwise import cosine_similarity
from nltk.tokenize import sent_tokenize
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
def split_semantic_paragraphs(result, video_duration=None, similarity_threshold=None):
    """
    Splits speech transcription (with timestamps) into semantically grouped paragraphs.

    Args:
        result: The output of the `speech_to_text` function, expected to be a dict with 'chunks',
                each containing 'text', 'timestamp'.
        video_duration: Duration of the video in seconds (used to set threshold dynamically).
        similarity_threshold: If provided, overrides the dynamic threshold.

    Returns:
        A list of tuples: (paragraph_text, (start_time, end_time))
    """
    # Determine similarity threshold based on video duration
    if similarity_threshold is None and video_duration is not None:
        duration_minutes = video_duration / 60
        if duration_minutes <= 10:
            similarity_threshold = 0.03
        elif duration_minutes <= 20:
            similarity_threshold = 0.025
        elif duration_minutes <= 30:
            similarity_threshold = 0.02
        elif duration_minutes <= 40:
            similarity_threshold = 0.015
        else:
            similarity_threshold = 0.01
    elif similarity_threshold is None:
        similarity_threshold = 0.02  # Default fallback

    print(f"Using similarity threshold: {similarity_threshold} \n\n\n\n\n\n\n")

    # Extract sentences and timestamps from result
    chunks = result['chunks']
    sentences = [chunk['text'] for chunk in chunks]
    timestamps = [chunk['timestamp'] for chunk in chunks]

    # Get embeddings
    embeddings = model.encode(sentences)

    paragraphs = []
    current_paragraph = [sentences[0]]
    current_times = [timestamps[0]]

    for i in range(1, len(sentences)):
        sim = cosine_similarity([embeddings[i]], [embeddings[i - 1]])[0][0]

        if sim >= similarity_threshold:
            current_paragraph.append(sentences[i])
            current_times.append(timestamps[i])
        else:
            start_time = current_times[0][0]
            end_time = current_times[-1][1]
            paragraphs.append((" ".join(current_paragraph), (start_time, end_time)))
            current_paragraph = [sentences[i]]
            current_times = [timestamps[i]]

    if current_paragraph:
        start_time = current_times[0][0]
        end_time = current_times[-1][1]
        paragraphs.append((" ".join(current_paragraph), (start_time, end_time)))

    return paragraphs

```

# Topic Modeling for Whole text

```python
pipe_topic = pipeline("text-generation", model="AmanPriyanshu/Dynamic-Topic-Modeling-Llama-3.2-1B-bnb-4bit")
```

```python
def topic_of_whole_text(text):
  messages = [
    {"role": "user", "content": text},
  ]
  # Identify the central topic or subject matter of the following text in a concise phrase:
  result = pipe_topic(messages)
  print(result)
  # Extract only the `content` fields
  contents = result[0]['generated_text'][1]['content']
  print(f"\n--- Whole text topic: ---\n{contents} ")
  return contents
```

# Topic Modeling for Each Paragraph

```python
def topic_model_of_paragraphs(paragraphs):
  paragraphs_topic = []
  for i, para in enumerate(paragraphs):
    messages = [
      {"role": "user", "content": para[0]},
    ]
    result_topic = pipe_topic(messages)
    # Extract only the `content` fields
    result_paragraph = result_topic[0]['generated_text'][1]['content']
    print(f"\n--- Paragraph {i+1} topic: ---\n{result_paragraph} ")
    print(f"\n--- Paragraph {i+1}: ---\n{para} ")
    paragraphs_topic.append(result_paragraph)
  return paragraphs_topic
```

# Text Summarization

```python
import nltk
nltk.download('punkt')
from nltk.tokenize import sent_tokenize

def summarize_and_split(text):
    sentences = sent_tokenize(text)
    return sentences  # List of sentences as strings
```

```python
def text_summarization(paragraph):
  messages = [
    {"role": "user", "content": f"Give me the summary of following text with at most 60 words: {paragraph}"},
  ]
  summary = pipe_llama(messages)
  print(summary)
  summary_text = summary[0]['generated_text'][1]['content']
  split = summarize_and_split(summary_text)
  return split, summary_text
```

# Key Phrase Extraction

```python
def key_phrase_extraction(paragraph):
  messages = [
    {"role": "user", "content": f"find up to 4 closest keyphrases or keywords to following text and just return the sequence of keywords(comma seperated, without numbering) and nothing else without any explanation: {paragraph}"},
  ]
  keyphrases = pipe_llama(messages)
  keyphrases = keyphrases[0]['generated_text'][1]['content']
  keyphrases = [item.strip() for item in keyphrases.split(',')]

  return keyphrases
```

# Query Search

```python
from sentence_transformers import SentenceTransformer, util

def find_keyword_and_semantic_match(keyword, sentences, query_sentence):
    # 1. Exact keyword match
    exact_matches = [s for s in sentences if f" {keyword} " in f" {s} "]

    # 2. Semantic similarity using SentenceTransformer
    model = SentenceTransformer('all-MiniLM-L6-v2')  # lightweight and fast
    sentence_embeddings = model.encode(sentences, convert_to_tensor=True)
    query_embedding = model.encode(query_sentence, convert_to_tensor=True)

    # Compute cosine similarities
    similarities = util.cos_sim(query_embedding, sentence_embeddings)[0]
    best_match_idx = similarities.argmax().item()
    best_match_sentence = sentences[best_match_idx]

    return exact_matches, best_match_sentence
```

# Find Related Scenes

```python
import cv2
import torch
import clip
from PIL import Image
import numpy as np
from IPython.display import display
from google.colab import files
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
model_clip, preprocess_clip = clip.load("ViT-B/32", device=device)
```

```python
def truncate_to_clip_limit(text, max_words=40):
    """
    CLIP tokenizer limit is 77 tokens. We'll truncate the text to at most 40 words as an approximation.
    """
    words = text.strip().split()
    return ' '.join(words[:max_words])
```

```python
def find_related_scene(query_text, video_path, target_start_time, target_end_time, chunk_duration=2.0):
  # Load CLIP
  # device = "cuda" if torch.cuda.is_available() else "cpu"
  # model, preprocess = clip.load("ViT-B/32", device=device)

  # Input
  # video_path = "REC-20250709184448.mp4"
  # query_text = "an apple and rabbit"
  # chunk_duration = 0.5  # seconds

  # Open video
  cap = cv2.VideoCapture(video_path)
  fps = cap.get(cv2.CAP_PROP_FPS)
  frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
  duration = frame_count / fps

  # Extract one frame from each 5s chunk
  scene_embeddings, scene_times, scene_images = [], [], []
  start = target_start_time
  while start < target_end_time:
      end = min(start + chunk_duration, duration)
      mid = (start + end) / 2
      cap.set(cv2.CAP_PROP_POS_MSEC, mid * 1000)
      ret, frame = cap.read()
      if ret:
          image = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
          pil_image = Image.fromarray(image)
          image_input = preprocess_clip(pil_image).unsqueeze(0).to(device)
          with torch.no_grad():
              emb = model_clip.encode_image(image_input)
          scene_embeddings.append(emb)
          scene_times.append((start, end))
          scene_images.append(pil_image)
      start += chunk_duration

  # Compute text embedding
  print(f"Query text: {query_text}")
  # text_input = clip.tokenize([query_text]).to(device)
  query_text_truncated = truncate_to_clip_limit(query_text)
  print(f"Query text (truncated): {query_text_truncated}")

  text_input = clip.tokenize([query_text_truncated]).to(device)
  with torch.no_grad():
      text_embedding = model_clip.encode_text(text_input)

  # Compare and find best match
  print(text_embedding, "Embedding")
  print(len(scene_embeddings), "Scene embedding ...\n\n\n\n")
  similarities = [torch.cosine_similarity(text_embedding, emb).item() for emb in scene_embeddings]
  best_idx = int(np.argmax(similarities))

  # Output
  print(f"Best matching segment: {scene_times[best_idx]}")
  display(scene_images[best_idx])  # Show matching frame
  return scene_images[best_idx], scene_images[0]
```

# Final Pipeline with base64 encode for file

```python
import os
import json
import base64
from pathlib import Path

def encode_file_base64(filepath):
    with open(filepath, "rb") as f:
        return base64.b64encode(f.read()).decode("utf-8")

def truncate_for_clip(text, max_words=60):
    words = text.split()
    return " ".join(words[:max_words])

def video_to_mindmap(video_path, output_dir="mindmap_output5"):
    os.makedirs(output_dir, exist_ok=True)

    # 1. Extract audio
    mp3_path = os.path.join(output_dir, "audio.mp3")
    duration = extract_audio_to_mp3(video_path, mp3_path)
    print(f"Duration: {duration} seconds\n\n\n\n\n\n")

    # 2. Transcribe
    result = speech_to_text(mp3_path)

    chunks = result['chunks']

    # 3. Paragraph segmentation
    paragraphs = split_semantic_paragraphs(result, video_duration=duration)

    # 4. Get root topic
    full_text = " ".join([p[0] for p in paragraphs])
    root_topic = topic_of_whole_text(full_text)

    mindmap = {"root_topic": root_topic, 'transcription': chunks, "nodes": []}

    # 5. Paragraph-level processing
    paragraph_topics = topic_model_of_paragraphs(paragraphs)
    for i, (text, (start, end)) in enumerate(paragraphs):
        print(f"\n⏳ Processing paragraph {i+1}/{len(paragraphs)}...")
        print(f"Paragraph: {text}")

        # Summarization
        bullet_summary, summary = text_summarization(text)
        print(f"Summary paragraph{i}: {summary}")

        # Keyphrase Extraction
        keywords = key_phrase_extraction(text)
        print(f"Keyphrases: {keywords}")

        # Named Entity Recognition
        named_entities = named_entity_recognition(summary)
        print(f"Named Entities: {named_entities}")

        # Emojis
        emojis = find_related_emojis(text)

        # Related Frame Image
        # safe_summary = truncate_for_clip(summary)
        best_frame, start_frame = find_related_scene(summary, video_path, start, end)
        best_frame_filename = f"paragraph_{i+1}_best_frame.jpg"
        start_frame_filename = f"paragraph_{i+1}_start_frame.jpg"
        best_frame_path = os.path.join(output_dir, best_frame_filename)
        start_frame_path = os.path.join(output_dir, start_frame_filename)
        best_frame.save(best_frame_path)
        start_frame.save(start_frame_path)

        # Text-to-Speech
        tts_filename = f"paragraph_{i+1}_summary.wav"
        tts_path = os.path.join(output_dir, tts_filename)
        text_to_speech(summary)
        os.rename("output.wav", tts_path)

        # Encode media to base64
        best_image_b64 = encode_file_base64(best_frame_path)
        start_image_b64 = encode_file_base64(start_frame_path)
        tts_b64 = encode_file_base64(tts_path)


        mindmap["nodes"].append({
            "timestamp": [start, end],
            "text": text,
            "summary": list(bullet_summary),
            "keywords": list(keywords),
            "named_entities": named_entities,
            "emojis": emojis,
            "best_image": best_image_b64,
            "start_image": start_image_b64,
            "topic": paragraph_topics[i],
            "tts": tts_b64
        })

    # Save output
    output_json = os.path.join(output_dir, "mindmap.json")
    with open(output_json, "w") as f:
        json.dump(mindmap, f, indent=2)

    print(f"\nMind map saved to {output_json}")
    return mindmap

```

```python
from fastapi import FastAPI, File, UploadFile
from fastapi.responses import JSONResponse
import uvicorn
import nest_asyncio
from pyngrok import ngrok
import shutil
import uuid
import os
import json
```

```python
from fastapi.middleware.cors import CORSMiddleware
from fastapi import BackgroundTasks
app = FastAPI()

app.add_middleware(
    CORSMiddleware,
    allow_origins=["*"],  # or restrict to your frontend domain
    allow_credentials=True,
    allow_methods=["*"],
    allow_headers=["*"],
)
@app.post("/generate-mindmap/")
async def generate_mindmap(background_tasks: BackgroundTasks, video: UploadFile = File(...)):
    video_id = str(uuid.uuid4())
    input_video_path = f"temp_{video_id}.mp4"

    with open(input_video_path, "wb") as buffer:
        shutil.copyfileobj(video.file, buffer)

    output_dir = f"output_{video_id}"
    os.makedirs(output_dir, exist_ok=True)

    def process_video():
        try:
            mindmap = video_to_mindmap(input_video_path, output_dir)
            with open(os.path.join(output_dir, "mindmap.json"), "w") as f:
                json.dump(mindmap, f)
        except Exception as e:
            with open(os.path.join(output_dir, "error.txt"), "w") as f:
                f.write(str(e))
        finally:
            os.remove(input_video_path)

    background_tasks.add_task(process_video)
    return {"status": "processing", "id": video_id}

@app.get("/mindmap-result/{video_id}")
def get_mindmap(video_id: str):
    output_dir = f"output_{video_id}"
    json_path = os.path.join(output_dir, "mindmap.json")
    error_path = os.path.join(output_dir, "error.txt")

    if os.path.exists(json_path):
        with open(json_path) as f:
            return json.load(f)
    elif os.path.exists(error_path):
        with open(error_path) as f:
            return {"error": f.read()}
    else:
        return {"status": "processing"}

```
